```python
import pandas as pd

# Cargar los datos desde el archivo Excel
file_path = 'C:/Users/VALE/Desktop/Tesis/Tesis Pedro Palominos/Código python/Modelo_info_historica/Archivos a usar en el modelo/solo var y asimetria_eleccionestaciones/Consol_Estacionelegidas_7_10 al 2_12.xlsx'  # Cambia esto al nombre de tu archivo Excel
sheet_name = "BD2PROM"  # Cambia esto si la hoja tiene otro nombre
df = pd.read_excel(file_path, sheet_name=sheet_name)

# Asegúrate de que las columnas tengan los nombres correctos
column_estacion = "Estacion"  # Cambia al nombre real de la columna para estaciones
column_bikes = "Bikes disponibles"  # Cambia al nombre real de la columna para las bicicletas

# Convertir la columna "Bikes disponibles" a numérica, manejando errores
df[column_bikes] = pd.to_numeric(df[column_bikes], errors="coerce")

# Eliminar filas donde "Bikes disponibles" sea NaN
df_clean = df.dropna(subset=[column_bikes])

# Agrupar por estación y calcular el promedio de "Bikes disponibles"
promedios = df_clean.groupby(column_estacion)[column_bikes].mean().reset_index()

# Renombrar la columna del promedio (opcional)
promedios.rename(columns={column_bikes: "Promedio Bikes"}, inplace=True)

# Exportar el resultado a un nuevo archivo Excel
output_file = "promedio_bikes_por_estacion.xlsx"
promedios.to_excel(output_file, index=False)

print(f"Archivo generado: {output_file}")

```
