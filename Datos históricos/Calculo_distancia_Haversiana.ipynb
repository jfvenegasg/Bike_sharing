```python
import pandas as pd
from math import radians, sin, cos, sqrt, atan2

# Función para calcular la distancia Haversine
def haversine(coord1, coord2):
    R = 6371  # Radio de la Tierra en km
    lat1, lon1 = radians(coord1[0]), radians(coord1[1])
    lat2, lon2 = radians(coord2[0]), radians(coord2[1])
    
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))
    return R * c  # Distancia en km

# Cargar las coordenadas desde un archivo Excel
file_path = 'C:/Users/VALE/Desktop/Tesis/Tesis Pedro Palominos/Código python/Modelo_info_historica/archivos_estaciones_elegidas20/Latitud_longitud_20.xlsx'# Cambia esto por el nombre de tu archivo
sheet_name = "Hoja2"  # Cambia esto si el archivo tiene otro nombre de hoja
df_coords = pd.read_excel(file_path, sheet_name=sheet_name)

# Crear el diccionario de coordenadas
coordenadas = {
    row['Estacion']: (row['Latitud'], row['Longitud'])
    for _, row in df_coords.iterrows()
    if pd.notna(row['Latitud']) and pd.notna(row['Longitud'])
}

# Crear la lista de distancias entre estaciones
data = []
for origen, coord_origen in coordenadas.items():
    for destino, coord_destino in coordenadas.items():
        if origen != destino:  # No calcular distancias de una estación consigo misma
            distancia = haversine(coord_origen, coord_destino)
            data.append([origen, destino, distancia, distancia])  # Agregar salida y entrada iguales

# Crear el DataFrame y exportarlo a Excel
df_distancias = pd.DataFrame(data, columns=["Estacion_Origen", "Estacion_Destino", "Costo_Salida", "Costo_Entrada"])
df_distancias.to_excel("distancias_estaciones.xlsx", index=False)

print("Archivo generado: distancias_estaciones.xlsx")
```
