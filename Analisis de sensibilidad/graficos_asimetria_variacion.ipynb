## Graficos Caso 1

Esta seccion grafica el mapa de calor de los balances correctos de la red para el caso 1

### Grafico suma de balances

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

archivo = 'Resultado_completoapromediocaso1.xlsx'
columnas = 'A:F'
df = pd.read_excel(archivo, sheet_name='mapa_calor_suma', header=1, usecols=columnas, engine='openpyxl')


df['Fechas'] = pd.to_datetime(df['Fechas'], errors='coerce')
df = df.dropna(subset=['Fechas'])  


df.set_index(df['Fechas'].dt.strftime('%d-%m-%Y'), inplace=True)
df.drop(columns=['Fechas'], inplace=True)


df = df.astype(float)

plt.figure(figsize=(12, 10))
sns.heatmap(df, annot=True, cmap="YlGnBu", linewidths=0.5, cbar_kws={'label': 'Balance'})
plt.title("Suma de balances totales correctos para un subconjunto de 20 estaciones de la red - Caso 1")
plt.ylabel("Fecha")
plt.xlabel("Variación de Asimetría")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
